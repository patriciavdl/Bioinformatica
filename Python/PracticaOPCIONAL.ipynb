#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

```python
resultado = []

for numero in valores:
    if numero % 2 == 0 and numero % 4 == 0:
        resultado.append("par-mult3:" + str(numero))
    elif numero % 2 == 0:
        resultado.append("par:" + str(numero))
    elif numero % 4 == 0:
        resultado.append("mult3:" + str(numero))
    else:
        resultado.append(str(numero))

print(resultado)
```

#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total". Marca un booleano como "aprobado" si el total es mayor a 50. Finalmente, crea un set con todos los nombres de los usuarios aprobados. 

```python
datos = [
    {"id":1,"nombre":"ana","puntos":(10,20,30)},
    {"id":2,"nombre":"luis","puntos":(5,15)},
    {"id":3,"nombre":"marta","puntos":(100,50,25)}
]
```

```python
aprobados = set()

for usuario in datos:
    usuario["nombre"] = usuario["nombre"].capitalize()
    usuario["total"] = sum(usuario["puntos"])
    usuario["aprobado"] = usuario["total"] > 50

    if usuario["aprobado"]:
        aprobados.add(usuario["nombre"])

print(datos)
print(aprobados)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
resultado = {}

for estudiante in estudiantes.values():
    nombre = estudiante["nombre"].upper()
    notas = estudiante["notas"]
    
    promedio = sum(notas) / len(notas)
    aprobado = promedio >= 20
    max_min = (max(notas), min(notas))
    
    resultado[nombre] = {
        "promedio": promedio,
        "aprobado": aprobado,
        "max_min": max_min
    }

print(resultado)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json

# Crear el archivo notas_alumnos.txt
with open("notas_alumnos.txt", "w") as archivo:
    archivo.write("Ana,10,20,30\n")
    archivo.write("Luis,5,15\n")
    archivo.write("Marta,100,50,25\n")

# Leer el archivo y crear la lista de diccionarios
alumnos = []

with open("notas_alumnos.txt", "r") as archivo:
    for linea in archivo:
        datos = linea.strip().split(",")
        
        nombre = datos[0].capitalize()
        notas = [int(nota) for nota in datos[1:]]
        promedio = sum(notas) / len(notas)
        
        alumno = {
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": promedio >= 20
        }
        
        alumnos.append(alumno)

# Guardar la lista en resultado.json
with open("resultado.json", "w") as archivo:
    json.dump(alumnos, archivo, indent=4)

print(alumnos)
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
import json
import csv

datos = {
    "u1": {"id": 1, "nombre": "ana", "roles": ["user", "admin"]},
    "u2": {"id": 2, "nombre": "luis", "roles": ["user"]},
    "u3": {"id": 3, "nombre": "marta", "roles": ["moderator", "user"]}
}

# Crear usuarios.json
with open("usuarios.json", "w") as archivo:
    json.dump(datos, archivo, indent=4)

# Abrir usuarios.json y leer los datos
with open("usuarios.json", "r") as archivo:
    usuarios = json.load(archivo)

# Crear usuarios.csv
with open("usuarios.csv", "w", newline="") as archivo:
    escritor = csv.writer(archivo)
    
    escritor.writerow(["id", "nombre", "roles"])
    
    for usuario in usuarios.values():
        nombre = usuario["nombre"].upper()
        roles = "|".join(usuario["roles"])
        
        escritor.writerow([usuario["id"], nombre, roles])
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
import json
import string

# Abrir el archivo
with open("fragmento-libro.txt", "r", encoding="utf-8") as archivo:
    texto = archivo.read()

# Pasar a minúsculas y eliminar signos
texto = texto.lower()
texto = texto.translate(str.maketrans("", "", string.punctuation))

# Separar en palabras
palabras = texto.split()

# Contar las palabras
frecuencias = {}

for palabra in palabras:
    if palabra in frecuencias:
        frecuencias[palabra] += 1
    else:
        frecuencias[palabra] = 1

# Ordenar de mayor a menor y quedarse con las 5 primeras
top_5 = sorted(frecuencias.items(), key=lambda x: x[1], reverse=True)[:5]

# Guardar en JSON
with open("frecuencias.json", "w", encoding="utf-8") as archivo:
    json.dump(dict(top_5), archivo, indent=4, ensure_ascii=False)
```

```python
print(top_5)
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
import csv

# Crear los tres archivos CSV
for nombre_archivo in ["ventas1.csv", "ventas2.csv", "ventas3.csv"]:
    with open(nombre_archivo, "w", newline="") as archivo:
        escritor = csv.writer(archivo)
        escritor.writerow(["producto", "cantidad"])
        escritor.writerow(["A", 10])
        escritor.writerow(["B", 5])

# Diccionario para acumular las ventas
ventas = {}

# Leer los tres archivos
for nombre_archivo in ["ventas1.csv", "ventas2.csv", "ventas3.csv"]:
    with open(nombre_archivo, "r") as archivo:
        lector = csv.DictReader(archivo)

        for fila in lector:
            producto = fila["producto"]
            cantidad = int(fila["cantidad"])

            if producto in ventas:
                ventas[producto] += cantidad
            else:
                ventas[producto] = cantidad

# Crear el archivo final
with open("total_ventas.csv", "w", newline="") as archivo:
    escritor = csv.writer(archivo)

    escritor.writerow(["producto", "cantidad"])

    for producto, cantidad in ventas.items():
        escritor.writerow([producto, cantidad])
```

```python
print(ventas)
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
import json

# Crear alumnos.json
datos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]

with open("alumnos.json", "w") as archivo:
    json.dump(datos, archivo, indent=4)

# Leer el archivo
with open("alumnos.json", "r") as archivo:
    alumnos = json.load(archivo)

# Crear el diccionario final
resultado = {}

for alumno in alumnos:
    nombre = alumno["nombre"]
    notas = alumno["notas"]

    promedio = sum(notas) / len(notas)

    resultado[nombre] = {
        "promedio": promedio,
        "max": max(notas),
        "min": min(notas),
        "aprobado": promedio >= 20
    }

print(resultado)
```

```python
from random import randint

numero = randint(1,101)
print(numero)

intentos = []
while True:
    intento = input("Escoje un numero del 1 al 100:")
    
    if int(intento) >= 1 and int(intento) <= 100:
        if len(intentos) == 0:
            if int(intento) == numero:
                intentos.append(intento)
                print("Acertaste a la primera ")
                break
            elif abs(int(intento) - numero) <=10:
                intentos.append(intento)
                print("CALIENTE")
                continue
            else: 
                intentos.append(intento)
                print("FRIO")
                continue
        else:
            if int(intento) == numero:
                intentos.append(intento)
                print("Acertaste con ", len(intentos),"intentos")
                break
            elif abs(numero - int(intentos[-1])) < abs(numero - int(intento)):
                intentos.append(intento)
                print("MAS FRIO")
            else:
                intentos.append(intento)
                print("MAS CALIENTE")
    else:
          print("FUERA DE LIMITES")
```

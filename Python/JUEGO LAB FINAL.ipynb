```python
import time
from IPython.display import Markdown, display, HTML

# Hemos realizado el trabajo de manera equitativa Almudena Jarillo y Patricia Villaplana
# empezamos usando "print" para explicar e introducir el contexto del juego
# usamos "import time" y "time.sleep()" para que deje un tiempo entre frases
# el \n es para saltar lineas 
# para que el texto salga en cursiva usamos "from IPython.display import Markdown, display" y pusimos "*" antes y despues del texto

def intro_laboratorio():
    display(HTML('<img src="labintro.jpg" width="600">'))
    print("–––––––––––––––––––––––––––––––––––––––––––––––––––")
    print("   🧪¡BIENVENIDO AL LABORATORIO ZENDAL GOMÉZ!🧪  ")
    print("–––––––––––––––––––––––––––––––––––––––––––––––––––")
    time.sleep(1.5)
    
    print("\nHola, buenas, entra, entra. Te vamos a mostrar como funciona nuestro laboratorio de virología. 🦠")
    time.sleep(2)
    
    print("Aquí analizamos todo tipo de muestras que vienen desde todo el mundo,")
    print("incluidas las mas peligrosas ⚠️ dado que tenemos un nivel de seguridad muy alto.")
    time.sleep(2)
    
    print("\nEspera... Oye, que es esto...")
    time.sleep(1.5)
    print("Uy... este tubo de ensayo debería estar tapado... 🧪🧪🧪")
    time.sleep(2)

    print("\n")
    display(Markdown("\n*☣️☣️DE REPENTE COMIENZA A SALIR HUMO VERDE DEL TUBO Y A GORGOTEAR☣️☣️*"))
    time.sleep(1)
    print("⚠️¡OH NOOOOO! ¡EL VIRUS SE HA ESCAPADO DEL TUBO!⚠️")
    time.sleep(2)
    display(Markdown("*🚨🚨el sistema de alarma se ha activado y ha provocado que el sistema de seguridad bloquee todas las puertas y ventanas, estais encerrados🚨🚨*"))
    time.sleep(2)


#con el "def" creamos una función en la cual va a estar guardado el juego
#primero llamamos la intro para que salga el texto de bienvenida para poder ejecutarla
def juego_laboratorio():
    intro_laboratorio()

    #"input" le da la opción al usuario para poder escribir lo que pida el juego (el querer ayudar o no)
    #".strip" elimina espacios en blanco accidentales que el usuario pueda escribir por error antes o despues del "si" o el "no"
    #".lower" evita confusiones entre mayusculas o minusculas al escribir "si" o "no"
    print("\n–––––––––––––––––––––––––––––––––––––––––––––––––––")
    respuesta = input("¡Necesitamos tu ayuda urgente! ¿Quieres ayudarnos a parar el virus? 🦠 (si/no): ").strip().lower()

    #si la respuesta no es igual a "si" (!=) el juego termina y si la respuesta es "si", el juego empieza
    #"return" interrumpe la función de "juego_laboratorio" y termina el juego
    if respuesta != "si":
        print("\n ¡No te vayas! *El virus se escapa y crea una catastrofe mundial...*")
        print("––– GAME OVER –––")
        return
        
    print("\n¡Genial! Rápido, ven y ponte la bata y los guantes. 😷🥼 Primero vamos a evaluar la carga viral.")
    time.sleep(2)

    #"carga_viral" crea una variable de la carga viral que tiene el virus, esto va a servir como contador de vida del jugador, si llegas a 100 se termina el juego porque no has podido contener el virus y si llegas a 0 has podido contener el virus + encontrar la cura y "ganas" el juego
    carga_viral = 50
    
    #"preguntas" es una lista (usa []) de diccionarios (usa {}), esto sirve para simplificar el codigo y que sea mas claro + corto ya que usar 5 bloques repetidos de if/elif/else seria lioso/muy extenso
    #en cada diccionario esta la pregunta ("titulo"), las tres opciones (A, B, C) con sus respesctivas calificaciones en el # y las respuestas escritas como "X_txt" y como cada respuesta cambia el valor de la carga viral ("X_val")
    preguntas = [
        {
            "titulo": "1. ¿Qué ámbito investigarías primero?",
            "opcion_A": "A) El material genético (ARN)", # CORRECTA (-15)
            "opcion_B": "B) Las proteínas virales",     # LENTA (+10)
            "opcion_C": "C) La entrada celular",       # GRAVE (+30)
            "A_txt": "\n--> Analizas el ARN de inmediato e identificas la secuencia base. 🧬 ¡Ganas tiempo!",
            "B_txt": "\n--> Intentar aislar proteínas requiere cristalización. Tarda demasiado y el virus se expande... ⌛️",
            "C_txt": "\n--> ¡Error grave! La presencia de un virus no implica que haya entrado en la celulas, podria encontrase en medio extracelular. 😵",
            "A_val": -15, "B_val": 10, "C_val": 30
        },
        {
            "titulo": "2. Una vez secuenciado el ARN, ¿qué técnica de diagnóstico rápido aplicas?",
            "opcion_A": "A) Realizar una prueba de anticuerpos IgG/IgM",       # GRAVE (+30)
            "opcion_B": "B) Diseñar una PCR en tiempo real (RT-qPCR)",         # CORRECTA (-15)
            "opcion_C": "C) Inocular y cultivar en huevos embrionados",         # LENTA (+10)
            "A_txt": "\n--> ¡ERROR GRAVE! Los anticuerpos tardan días en generarse. En un brote activo no detectas nada a tiempo. ⌛️🚨",
            "B_txt": "\n--> ¡Perfecto! La RT-qPCR amplifica el material genético e identifica el patógeno en minutos. 🧬🔎",
            "C_txt": "\n--> El cultivo viral tradicional funciona, pero necesita días de incubación. Se pierde tiempo muy valioso... ⌛️",
            "A_val": 30, "B_val": -15, "C_val": 10
        },
        {
            "titulo": "3. Para evitar que el virus siga infectando células, ¿en qué estructura te enfocas?",
            "opcion_A": "A) Inactivar la envoltura con un detergente iónico fuerte", # GRAVE (+30)
            "opcion_B": "B) Bloquear la proteasa celular que procesa la cápside",     # LENTA (+10)
            "opcion_C": "C) Bloquear la proteína Espícula (Spike) de superficie",     # CORRECTA (-15)
            "A_txt": "\n--> ¡PERO QUÉ HACES! Un detergente iónico destruye también la membrana de las células sanas del ensayo. 😱😱😱",
            "B_txt": "\n--> Funciona como vía secundaria, pero bloquear proteasas humanas causa toxicidad y ralentiza el proceso. 🐌",
            "C_txt": "\n--> ¡Exacto! Al unirte a la proteína Spike, impides que el virus encaje en el receptor ACE2. 🎉",
            "A_val": 30, "B_val": 10, "C_val": -15
        },
        {
            "titulo": "4. El virus ataca la muestra sintética de prueba. ¿Qué inhibidor aplicas?",
            "opcion_A": "A) Inhibidor de la polimerasa (frena replicación)", # CORRECTA (-15)
            "opcion_B": "B) Antibiótico de amplio espectro",                # GRAVE (+30)
            "opcion_C": "C) Congelar la muestra a -80°C",                   # LENTA (+10)
            "A_txt": "\n--> ¡Muy bien! Detienes la capacidad del virus para replicarse. 👏🏼",
            "B_txt": "\n--> ¡ESTÁS MEGALOCO! Los antibióticos matan bacterias, NO virus. Has perdido tiempo valioso. 🫪😟",
            "C_txt": "\n--> La ultracongelación frena el virus, pero no lo destruye. En cuanto suba la temperatura volverá a actuar. 🥶🧊",
            "A_val": -15, "B_val": 30, "C_val": 10
        },
        {
            "titulo": "5. ¡Fase final! Tienes que diseñar la terapia curativa definitiva. ¿Cuál eliges?",
            "opcion_A": "A) Diseñar un suero con plasma de paciente sin purificar", # LENTA (+10)
            "opcion_B": "B) Administrar una dosis elevada de interferón sintético",      # GRAVE (+30)
            "opcion_C": "C) Sintetizar anticuerpos monoclonales neutralizantes",   # CORRECTA (-15)
            "A_txt": "\n--> El plasma podria ayudar, pero requiere pruebas de compatibilidad y filtrado lento lo cual retrasaria tu avance... 🐌",
            "B_txt": "\n--> ¡PELIGRO! Provocas un excesivo numero de citoquinas que provoca que tu modelo de estudio se vaya fuera de control. ⚠️",
            "C_txt": "\n--> ¡Tecnica definitiva! Los anticuerpos monoclonales se unen específicamente al virus y lo neutralizan por completo, has salvado al mundo. 🎉🌍",
            "A_val": 10, "B_val": 30, "C_val": -15
        }
    ]
    #"for" crea un bucle para en el diccionario creado vaya procesando las preguntas en orden consecutivo
    #usamos "p" en vez de "preguntas" en el bucle para simplificar el codigo y no tener que estar escribiendo "preguntas" cada vez que lo mencionamos
    for p in preguntas:
        #verificamos si antes de hacer la pregunta la carga viral ha sobrepasado el limite permitido 
        if carga_viral >= 100:
            break
            
    #"f-string" imprime el valor de la variable "carga_viral"
    #"p" abre el diccionario y luego entre "[]" sacamos lo que queremos extraer del diccionario
        print("\n–––––––––––––––––––––––––––––––––––––––––––––––––")
        print(f" CARGA VIRAL: {carga_viral}% ")
        print("–––––––––––––––––––––––––––––––––––––––––––––––––")
        print(p["titulo"])
        print(p["opcion_A"])
        print(p["opcion_B"])
        print(p["opcion_C"])

        #jugador escoge su intento
        intento = input("\nElige una opción (A, B o C): ").strip().upper()

        #"while" elimina cualquier opción que no sea una de las indicadas (A, B o C) y te devuelve una frase y el intento de volver a elegir
        while intento not in ["A", "B", "C"]:
            print("[!] Opción no válida. Por favor, elige A, B o C.")
            intento = input("Elige una opción (A, B o C): ").strip().upper()

        #segun la opción elegida te devuelve su texto asociado y cambia el valor de la variable "carga_viral"
        #"+=" simplifica el codigo y guarda el nuevo valor de la variable "carga_viral"
        if intento == "A":
            print(p["A_txt"])
            carga_viral += p["A_val"]
        elif intento == "B":
            print(p["B_txt"])
            carga_viral += p["B_val"]
        elif intento == "C":
            print(p["C_txt"])
            carga_viral += p["C_val"]

        time.sleep(1.5)

        #si en cualquier momento el valor de la variable "carga_viral" es igual a o menor que 0 –> jugador ha ganado directamente
        if carga_viral <= 0:
            carga_viral = 0
            display(HTML('<img src="cargaviralzero.jpg" width="600">'))
            print("\n👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼")
            print("🎉¡¡¡CARGA VIRAL REDUCIDA A 0%!!!🎉")
            print("🤗¡Lo logramos! Has neutralizado la amenaza a tiempo.🤗")
            print("🌍¡HAS SALVADO EL LABORATORIO Y AL MUNDO!🌍")
            print("👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼👏🏼")
            return

    
    #si la carga viral es igual a o mayor que 100 -> jugador ha perdido 
    #"f-string" te devuelve el valor de la variable "carga_viral" con el cual has "muerto"
    if carga_viral >= 100:
        print("\nOh no... parece ser que te has quedado sin tiempo...")
        time.sleep(2)
        display(HTML('<img src="cargaviralcien.jpg" width="600">'))
        print("\n☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️")
        print(f"¡LA CARGA VIRAL HA SUPERADO EL LÍMITE DE SEGURIDAD! ({carga_viral}%)")
        print("💥💀HAS MUERTO / EL VIRUS HA GANADO. El virus escapa del laboratorio y causa contaminación masiva.💀💥")
        print("––– ☠️ GAME OVER ☠️ –––")
        print("☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️☠️")

    #si ya finalizada las 5 preguntas la carga sigue entre 1 y 99 -> el tiempo se ha agotado
    #"f-string" te devuelve el valor de la variable "carga_viral" con el cual te has quedado
    elif carga_viral > 0:
        print("\nOh no... parece ser que te has quedado sin tiempo...")
        time.sleep(2)
        display(HTML('<img src="cargaviraltiempo.jpg" width="600">'))
        print("\n⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️")
        print(f"¡TE HAS QUEDADO SIN TIEMPO! (Carga viral restante: {carga_viral}%)")
        print("🐌 Hemos agotado los 5 protocolos de emergencia del sistema. 🐌")
        print("⌛️ Aunque la carga viral no te ha matado, no has logrado neutralizar el virus a tiempo. ⌛️")
        print("🪖 El protocolo militar ha sellado el laboratorio contigo dentro para evitar un brote exterior. 🪖")
        print("––– SE HA ACABDO EL TIEMPO / HAS PERDIDO –––")
        print("⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️⏱️")


#llamas a el juego
juego_laboratorio()
```
